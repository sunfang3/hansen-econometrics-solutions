> **写给只学过李子奈/陈强的同学：** 本章在**误差正态**假设 $e\mid X\sim N(0,\sigma^2 I)$ 下，推出 $\hat\beta$、残差、$s^2$、$t$、$F$ 的**精确有限样本分布**。要点：正态性是"花钱买精确分布"——不花钱（去掉正态）就只能要 Ch.7 的**渐近**分布。
> 四个核心工具：(1) 正态的线性变换仍正态（→ $\hat\beta$、$\hat e$ 的分布）；(2) 联合正态下"不相关 ⟺ 独立"（→ $\hat\beta\perp\hat e$，因 $X'M=0$）；(3) 幂等二次型 ⟹ $\chi^2$（→ $(n-k)s^2/\sigma^2\sim\chi^2_{n-k}$）；(4) $t=N/\sqrt{\chi^2/\nu}$、$F=(\chi^2/q)/(\chi^2/\nu)$。
> **警示：** 精确 $t$ 分布**只**对**同方差** SE 成立；稳健 $t$ 的大样本合法性要等 Ch.7。


```python
import numpy as np
rng = np.random.default_rng(0)
r = 5
Z = rng.normal(size=(200000, r))
Q = np.sum(Z**2, axis=1)
print("E[Q] sample", Q.mean(), "theory", r)
print("Var[Q] sample", Q.var(ddof=1), "theory", 2*r)
```

## 理论结论的蒙特卡洛验证（正态回归，无需外部数据）

以下单元格核对 ch05 的精确分布：$\chi^2$ 矩、拟合值方差 $\sigma^2 h_{ii}$、$\hat\beta$ 与 $s^2$ 独立、$\mathrm{var}(s^2)$ 与 CRLB。可独立运行。

```python
import numpy as np
rng = np.random.default_rng(5)

# 正态回归设定（误差 i.i.d. N(0,σ²)），条件于 X 做大量重复
n, k, reps = 60, 4, 60000
Xc = rng.standard_normal((n, k-1))
X = np.c_[Xc, np.ones(n)]
beta = np.array([1.0, -0.5, 0.7, 2.0])
sigma2 = 4.0
XtXinv = np.linalg.inv(X.T @ X)
h = np.diag(X @ XtXinv @ X.T)

Bh = np.zeros((reps, k)); Yhat3 = np.zeros(reps); s2 = np.zeros(reps); chic = np.zeros(reps)
for r in range(reps):
    e = rng.standard_normal(n) * np.sqrt(sigma2)
    Y = X @ beta + e
    bh = XtXinv @ (X.T @ Y)
    eh = Y - X @ bh
    Bh[r] = bh
    Yhat3[r] = X[3] @ bh
    s2[r] = (eh @ eh) / (n - k)
    chic[r] = (n - k) * s2[r] / sigma2

# Ex 5.1: χ²_r 的均值=r、方差=2r
Q = np.sum(rng.standard_normal((200000, 5))**2, axis=1)
print(f"[5.1]  χ²_5: E={Q.mean():.3f}(应5)  Var={Q.var():.3f}(应10)")

# Ex 5.5: 拟合值方差 = σ² h_ii
print(f"[5.5]  var(Ŷ_i)={Yhat3.var():.5f} ≈ σ²·h_ii={sigma2*h[3]:.5f}")

# Ex 5.6/5.7: 正态下 β̂ 与 s² 不相关 ⇒ 独立
print(f"[5.6/7] cov(β̂,s²)≈0:", np.round(np.cov(Bh.T, s2)[:k, -1], 4))

# Ex 5.12: (n-k)s²/σ² ~ χ²_{n-k}；var(s²)=2σ⁴/(n-k) > CRLB=2σ⁴/n
print(f"[5.12] (n-k)s²/σ²: 均值={chic.mean():.2f}(={n-k})  方差={chic.var():.2f}(={2*(n-k)})")
print(f"       var(s²)={s2.var():.5f}=2σ⁴/(n-k)={2*sigma2**2/(n-k):.5f} > CRLB 2σ⁴/n={2*sigma2**2/n:.5f}")
```
